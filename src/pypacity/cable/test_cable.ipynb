```python
import pandas as pd
from cable import Cable
```

```python
filename = './cable_db.csv'
cable_db = pd.read_csv(filename, sep=';')

cable_db
```

```python
NSELECT = 2

drake = Cable()
drake.set_cable(NSELECT, conductor='DRAKE')

print('ID:', drake.ID)
print('D [mm]:', drake.D)
print('RLO [ohm/m]:', drake.RLO)
print('RHI [ohm/m]:', drake.RHI)
print('HEATCAP [W.s/(m.deg C)]:', drake.HEATCAP)
```

```python
for cable_id in cable_db['ID']:
    cable = Cable()
    cable.set_cable(NSELECT, conductor=cable_id)
    print(cable.ID, cable.D, cable.RLO, cable.RHI)
```

```python
custom = Cable()
custom.set_cable(NSELECT, conductor='DRAKE')

custom.ID = 'DRAKE_CUSTOM'
custom.D = 30.0
custom.TCDRMAX = 120.0
custom.RLO = 0.08000 / 1000.0
custom.RHI = 0.09500 / 1000.0
custom.HEATCAP = custom.HEATCORE + custom.HEATOUT

print('ID:', custom.ID)
print('D [mm]:', custom.D)
print('TCDRMAX [deg C]:', custom.TCDRMAX)
print('RLO [ohm/m]:', custom.RLO)
print('RHI [ohm/m]:', custom.RHI)
```
